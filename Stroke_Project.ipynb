```python
import pandas as pd

df = pd.read_csv('healthcare-dataset-stroke-data.csv')
print(df.shape)
print(df.head())
print(df.isnull().sum())
print(df['stroke'].value_counts())
```

```python
import matplotlib.pyplot as plt

df['stroke'].value_counts().plot(kind='bar', color=['steelblue','tomato'])
plt.title('Stroke vs No Stroke')
plt.xticks([0,1], ['No Stroke','Stroke'], rotation=0)
plt.show()
```

```python
df['bmi'] = df['bmi'].fillna(df['bmi'].mean())
df = df.drop('id', axis=1, errors='ignore')
print(df.shape)
```

```python
df = pd.get_dummies(df, columns=['gender','ever_married','work_type',
                                  'Residence_type','smoking_status'])

df = df.drop('gender_Other', axis=1)
print(df.shape)
print(df.columns.tolist())
```
